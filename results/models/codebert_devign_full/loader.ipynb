```python
from pathlib import Path
import torch
from transformers import AutoModelForSequenceClassification, AutoTokenizer

# Resolve the model directory from this notebook, even when the kernel starts elsewhere.
MODEL_DIR = Path.cwd()
if not (MODEL_DIR / "config.json").exists():
    MODEL_DIR = Path(__file__).resolve().parent if "__file__" in globals() else MODEL_DIR

print(f"Loading fine-tuned CodeBERT model from {MODEL_DIR}...")
tokenizer = AutoTokenizer.from_pretrained(
    MODEL_DIR,
    local_files_only=True,
    use_fast=True,
 )
model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_DIR,
    local_files_only=True,
    use_safetensors=True,
 )

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
model.eval()
print(f"Model loaded on {device}.")

def predict_code(code: str) -> tuple[int, str]:
    """Return the predicted class ID and label for one code snippet."""
    if not code.strip():
        raise ValueError("Code snippet cannot be empty.")
    inputs = tokenizer(
        code,
        return_tensors="pt",
        truncation=True,
        max_length=512,
    ).to(device)
    with torch.no_grad():
        prediction_id = torch.argmax(model(**inputs).logits, dim=-1).item()
    label = model.config.id2label.get(prediction_id, str(prediction_id))
    return prediction_id, label

print("Ready. Call predict_code(code) or set RUN_INTERACTIVE = True.")
RUN_INTERACTIVE = False

if RUN_INTERACTIVE:
    while True:
        user_code = input("\nEnter code snippet (or 'exit'): ")
        if user_code.strip().lower() == "exit":
            break
        if user_code.strip():
            prediction_id, label = predict_code(user_code)
            print(f"Prediction: {label} (label ID: {prediction_id})")
```
